```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(
    lang="en",
    enable_mkldnn=False,
    use_doc_orientation_classify=True,
    use_doc_unwarping=True,
    use_textline_orientation=True,
)
```

```python
from paddleocr import FormulaRecognitionPipeline

formula_pipeline_l = FormulaRecognitionPipeline(
    formula_recognition_model_name="PP-FormulaNet_plus-L",
    device="cpu",
    use_layout_detection=False,
    enable_mkldnn=False,
)
```

```python
from pathlib import Path

file_path = Path("../test_data/simple_math_formula_1.png")

if not file_path.exists():
    raise FileNotFoundError(
        f"File not found: {file_path.resolve()}"
    )

print("File:", file_path.resolve())
```

```python
result = formula_pipeline_l.predict(str(file_path))
```

```python
for page_no, res in enumerate(result, start=1):

    data = res.json["res"]

    print(f"\n========== PAGE {page_no} ==========")

    formulas = data.get("formula_res_list", [])

    print(f"Formula count: {len(formulas)}")

    for i, formula in enumerate(formulas, start=1):
        print(f"\nFormula {i}")
        print("LaTeX:", formula.get("rec_formula"))
```
