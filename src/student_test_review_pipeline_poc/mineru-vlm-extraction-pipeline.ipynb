### testing mineru library for parsing and extracting math formula


```python
from pathlib import Path
import json

from mineru.parser import parse
```

```python
PDF_PATH = Path(
    # "../test_data/maths_handwritten_notes.pdf"
    "../test_data/maths-ch-8-docs/ch_8_Q_8_test.pdf")

print("Exists:", PDF_PATH.exists())
print("File:", PDF_PATH)
```

```python
result = parse(
    PDF_PATH,
    tier="standard",
)

print("Parsing completed successfully.")
print(type(result))
```

### Get & Save Markdown

```python
markdown = result.markdown()

print(markdown)
```

```python
OUTPUT_DIR = Path("mineru_output")
OUTPUT_DIR.mkdir(exist_ok=True)

markdown_path = OUTPUT_DIR / "ch_8_Q_8_test.md"

markdown_path.write_text(
    markdown,
    encoding="utf-8",
)

print(f"Saved: {markdown_path}")
```

### Get structured JSON

```python
data = result.to_dict()

print(type(data))
print(data.keys() if isinstance(data, dict) else "Not a dictionary")
```

```python
json_path = OUTPUT_DIR / "ch_8_Q_8_test.json"

json_path.write_text(
    json.dumps(
        data,
        indent=2,
        ensure_ascii=False,
    ),
    encoding="utf-8",
)

print(f"Saved: {json_path}")
```

### Pretty-print the structure

```python
from pprint import pprint

pprint(data, depth=5)
```
